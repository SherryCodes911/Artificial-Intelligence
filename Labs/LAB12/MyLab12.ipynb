```python
#input=np.array([[2,3,4],[3,4,5],[5,6,7]])
input=np.array([[2, 3, 4], [3, 4, 5], [5, 6, 7]])
weight={
    "node_01": np.array([1,1,1]),
    "node_02": np.array([1,2,-1]),
    "node_03": np.array([3,-2,1]),
    "node_04": np.array([1,-1,-1]),

    "node_10": np.array([2,1,2,1]),
    "node_20": np.array([1,2,-1,2]),
    "node_30": np.array([1,-2,1,-2]),
    "node_40": np.array([1,-1,-1,1]),

    "output": np.array([1,3,2,1])
}
def relu(input):
    return max(0,input)

def meth(input,weight):
    #inputs
    node_01_input=(input*weight["node_01"]).sum()
    node_02_input=(input*weight["node_02"]).sum()
    node_03_input=(input*weight["node_03"]).sum()
    node_04_input=(input*weight["node_04"]).sum()
    #outputs
    node_01_output=relu(node_01_input)
    node_02_output=relu(node_02_input)
    node_03_output=relu(node_03_input)
    node_04_output=relu(node_04_input)
    #hidden values 1
    hidden_val_1=np.array([node_01_output, node_02_output, node_03_output, node_04_output])
    
    #inputs
    node_10_input=(hidden_val_1*weight["node_10"]).sum()
    node_20_input=(hidden_val_1*weight["node_20"]).sum()
    node_30_input=(hidden_val_1*weight["node_30"]).sum()
    node_40_input=(hidden_val_1*weight["node_40"]).sum()
    
    node_10_output=relu(node_10_input)
    node_20_output=relu(node_20_input)
    node_30_output=relu(node_30_input)
    node_40_output=relu(node_40_input)
    #hidden values 2
    hidden_val_2=np.array([node_10_output, node_20_output, node_30_output, node_40_output])

    return (hidden_val_2 * weight["output"]).sum()
```
