```python
#task1:Working with Regular Expressions
import re
pattern=r'^[A-Z][\s\S]*\.$'

text="Hello world"
match=re.match(pattern, text)
print(bool(match))
```

```python
pattern=r'^\d{5}$'

zip_code="78100"
match=re.match(pattern, zip_code)
print(bool(match))
```

```python
#task02: Bag of Words
import re
from collections import Counter

dialogue="""
Character A: "The galaxy is in turmoil. The rebels are fighting
to overthrow the empire, and we need to find a way to communicate
with the leaders. We can't let the forces of darkness take over our
world."
Character B: "I agree. The resistance is our only hope. But how
can we gain support from the people? We need a strategy to win their
hearts and minds, to unite them against the empire."
Character A: "We must focus on the key battles and disrupt their communication lines.
We also need to create a powerful message that inspires hope. If we
can sway public opinion, we can win this war."
"""
```

```python
dialogue=dialogue.lower()
dialogue=re.sub(r'[^a-z\s]', '', dialogue)
dialogue
```

```python
#BoW
words=dialogue.split()
words
```

```python
#frequency of each word
word_freq=Counter(words)
word_freq
```

```python
#topic
stopwords={"the", "and", "we", "to", "a", "of", "our", "is", "can", "on", "in", "this", "but", "from", "with", "how", "if", "also"}
keywords=[word for word in words if word not in stopwords]
topic_freq=Counter(keywords)
topic_freq
```

```python
print("Top Keywords regarding topic:", topic_freq.most_common(5))
```
